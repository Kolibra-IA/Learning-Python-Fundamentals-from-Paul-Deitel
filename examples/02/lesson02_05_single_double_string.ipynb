# Single and Double Quoted Strings

#### ' : Single Quote.

#### " : Double Quote.

#### \ : Escape Character.

#### \n : Escape Sequence for a New Line.

#### \\' : Escape Sequence for a Single Quote.

#### \\" : Escape Sequence for a Double Quote.

## ' : Single Quote

This is used to represent a string in Python.

```python
x = 'Ali is a boy';
x
```

## " : Double Quote

Alternatively, double strings can be used to represent a string in Python.

```python
x = "Ali is a boy";
x
```

## \ : Escape Character

This is used to introduce a special meaning in Python. The escape character and the next character often form an escape sequence.

###### \

Also, a single statement can be split into multiple lines with an escape character in ipython - not in Jupyper Notebook; just had to explain it here for reference.

```python
x = 'This is awesome. \ Yes, baby!'

print(x);
```

## \n : Escape Sequence for a New Line.

This represent a new line in Python.

```python
x = 'This is a new line. \n You can check it out here now!';

print(x);

x
```

## \\' : Escape Sequence for a Single Quote.

This is used to introduce a single quote within opening and closing single quotes, so the behavior is not unexpected.

```python
x = 'Ali\'s shoes';

x
```

## \\" : Escape Sequence for a Double Quote.

This is used to introduce a single quote within opening and closing single quotes, so the behavior is not unexpected.

```python
x = "Ali's shoes are \"in\" the closet";

x
```
