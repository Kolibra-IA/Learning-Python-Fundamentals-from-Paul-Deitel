#### Exponentiation

```python
x = 9**3; 

print(x);
```

#### Multiplication: *



```python
x = 3*3;

print(x);
```



#### Division with Float: /


```python
x = 9/3;

print(x);
```



#### Division without Float (Literallly the Floor Value) : //


The result is the largest whole number that is not greater than the math result of the operation. 

For example, -13//4's result is -3.25; however, the largest whole number that is not greater than -3.25 is not -3 but -4; think about that slowly but calmly.

```python
x = 9//3;

print(x);
```



#### Addition: +



```python
x = 9+3;

print(x);
```



#### Substraction: -

```python
x = 9-3;

print(x);
```
